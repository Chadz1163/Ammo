### 创建一个基础的Agent

```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"{city}总是晴天!"

agent = create_agent(
    model="deepseek-reasoner",
    tools=[get_weather],
    system_prompt="你是一个天气助手",
)

# Run the agent
agent.invoke(
    {"messages": [{"role": "user", "content": "现在上海的天气怎么样？"}]}
)
```

#### ***调用***  
必须调用聊天模型才能生成输出。有三种主要的调用方法，每种方法都适合不同的用例

1. 调用——`invoke()`

```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage

conversation = [
    {"role": "system", "content": "You are a helpful assistant that translates English to Chinese."},
    {"role": "user", "content": "Translate: I love programming."},
    {"role": "assistant", "content": "J'adore la programmation."},
    {"role": "user", "content": "Translate: I love building applications."}
]

response = model.invoke(conversation)
print(response)  # AIMessage("J'adore créer des applications.")
```

2. 流——`stream()`  

大多数模型可以在生成过程中流式传输输出内容。通过逐步显示输出，流式传输显著提升了用户体验，尤其是对更长的响应

```python
full = None  # None | AIMessageChunk
for chunk in model.stream("What color is the sky?"):
    full = chunk if full is None else full + chunk
    print(full.text)

# The
# The sky
# The sky is
# The sky is typically
# The sky is typically blue
# ...

print(full.content_blocks)
# [{"type": "text", "text": "The sky is typically blue..."}]
```

```python
response = model.stream("上海今天天气怎么样?")
full = None  # None | AIMessageChunk
for chunk in response:
    # full = chunk if full is None else full + chunk
    print(chunk.text, end="", flush=True)
```

### 批处理——`batch()`   
批量处理一组独立请求到模型可以显著提升性能并降低成本，因为处理可以并行完成：

```python
responses = model.batch([
    "为什么天空是蓝色的?",
    "飞机是如何飞行的?",
    "什么是量子计算?"
])
for response in responses:
    print(response)
```

默认情况下，batch（） 只返回整个批次的最终输出。如果你想在每个输入生成完成时接收输出，可以用 batch_as_completed（）） 流式传输结果：

```python
for response in model.batch_as_completed([
    "为什么天空是蓝色的?",
    "飞机是如何飞行的?",
    "什么是量子计算?"
]):
    print(response)
```

## 工具调用  
模型可以请求调用执行诸如从数据库获取数据、搜索网页或运行代码等任务的工具。工具是以下组合：
1. 一个模式，包括工具名称、描述和/或参数定义（通常是 JSON 模式）
2. 一个函数或协程执行。

```python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
    """Get the weather at a location."""
    return f"It's sunny in {location}."


model_with_tools = model.bind_tools([get_weather])  

response = model_with_tools.invoke("What's the weather like in Boston?")
for tool_call in response.tool_calls:
    # View tool calls made by the model
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```

### 结构化输出   
模型可以被请求以符合特定模式的格式提供响应。这有助于确保输出易于解析并在后续处理中使用。   
LangChain 支持多种模式类型和多种方法来强制结构化输出。

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    """A movie with details."""
    title: str = Field(..., description="电影的主题")
    year: int = Field(..., description="电影的上映年份")
    director: str = Field(..., description="电影的导演")
    rating: float = Field(..., description="电影的评分")

model_with_structure = model.with_structured_output(Movie)
response = model_with_structure.invoke("提供关于电影盗梦的详细信息")
print(response)  # Movie(title="Inception", year=2010, director="Christopher Nolan", rating=8.8)
```

### 提示缓存

许多提供商提供即时缓存功能，以降低重复处理相同令牌时的延迟和成本。这些特征可以是隐式的，也可以是显式的：   
- 隐式提示缓存：如果请求进入缓存，提供者会自动将节省的成本转嫁给用户。示例：OpenAI和Gemini（Gemini 2.5及以上版本）。
- 显式缓存：提供商允许您手动指定缓存点以增强控制或保证节省成本。示例：ChatOpenAI（通过）、Anthropic的AnthropicPromptCachingMiddleware及cache_control选项、AWS Bedrock、Gemini。prompt_cache_key
