### 使用langchain-deepseek包调用deepseek模型

```python
import os
import langchain_deepseek
from dotenv import load_dotenv

# 加载环境变量
load_dotenv(".env")

print(langchain_deepseek.__version__)


llm_deepseek = langchain_deepseek.ChatDeepSeek(
    model="deepseek-reasoner",
    api_key=os.environ.get('DEEPSEEK_API_KEY'),
    api_base=os.environ.get('DEEPSEEK_BASE_URL'),
    max_tokens=1024,
    temperature=1.3,
)


response = llm_deepseek.invoke("你好，你是谁？")

print(type(response))
print(response)
```

### 使用 openai 包调用 deepseek 模型

```python
import os
import langchain_openai
from dotenv import load_dotenv

# 加载环境变量
load_dotenv(".env")


llm_deepseek = langchain_openai.ChatOpenAI(
    model="deepseek-reasoner",
    api_key=os.environ.get('DEEPSEEK_API_KEY'),
    base_url=os.environ.get('DEEPSEEK_BASE_URL'),
    max_tokens=1024,
    temperature=1.3,
)


response = llm_deepseek.invoke("你好，你是谁？")

print(type(response))
print(response)
```

***两者的区别在于，使用deepseek提供的包能够获得更多的信息，能够更加定制化的处理deepseek模型的输出。***


### Temperature 设置  (deepseek)
`temperature` 参数默认为 1.0  
- 我们建议您根据如下表格，按使用场景设置 temperature  

| 场景 | 温度/随机性 |
| --- | --- |
| 代码生成/数学解题 | 0.0 |
| 数据抽取/分析 | 1.0 |
| 通用对话 | 1.3 |
| 翻译 | 1.3 |
| 创意类写作/诗歌创作 | 1.5 |

## LangChain QuickStart

### 创建一个基础的Agent

```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"{city}总是晴天!"

agent = create_agent(
    model="deepseek-reasoner",
    tools=[get_weather],
    system_prompt="你是一个天气助手",
)

# Run the agent
agent.invoke(
    {"messages": [{"role": "user", "content": "现在上海的天气怎么样？"}]}
)
```
