## 第二章：Models（模型）- LangChain的基石

### 2.1 为什么Models是基石？

在LangChain中，**Model（模型）**是一切的起点。它们是：
- 🧠 **大脑**：负责理解和生成文本
- 🔌 **接口**：连接不同AI提供商的统一标准
- 🎯 **核心**：Agents、Chains、RAG等所有功能的基础

### 2.2 LangChain的模型标准化接口

LangChain最大的优势是**统一了不同提供商的API**，无论使用OpenAI、Anthropic、Google还是本地模型，代码基本相同。

#### 核心接口方法

| 方法 | 作用 | 使用场景 |
|-----|------|---------|
| `invoke()` | 单次完整调用 | 获取一次性完整响应 |
| `stream()` | 流式输出 | 实时显示生成内容 |
| `batch()` | 批量处理 | 同时处理多个请求 |
| `bind_tools()` | 绑定工具 | 让模型能够调用函数 |

## 第三章：Messages（消息）- 模型间的通信协议

### 3.1 什么是Messages？

在LangChain中，**Messages**是模型之间交流的"语言"。就像人类对话有不同的角色（说话者、听众），LangChain也定义了不同类型的消息。

### 3.2 四种核心消息类型

```
┌──────────────────┐
│ SystemMessage    │ ← 给模型的系统指令（设定角色、规则）
├──────────────────┤
│ HumanMessage     │ ← 用户的输入
├──────────────────┤
│ AIMessage        │ ← AI的响应
├──────────────────┤
│ ToolMessage      │ ← 工具执行的结果
└──────────────────┘
```

### 3.3 详解每种消息类型

````python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage

print("="*60)
print("1. SystemMessage - 系统指令")
print("="*60)
system_msg = SystemMessage(content="你是一个专业的Python教师，解释要简单易懂")
print(f"类型: {system_msg.type}")
print(f"内容: {system_msg.content}")
print("作用: 设定AI的角色、行为规则、输出格式等")
print()

print("="*60)
print("2. HumanMessage - 用户输入")
print("="*60)
human_msg = HumanMessage(content="什么是列表推导式？")
print(f"类型: {human_msg.type}")
print(f"内容: {human_msg.content}")
print("作用: 代表用户的问题或指令")
print()

print("="*60)
print("3. AIMessage - AI响应")
print("="*60)
# 通常不需要手动创建，模型会返回
ai_msg = AIMessage(content="列表推导式是Python中创建列表的简洁方式...")
print(f"类型: {ai_msg.type}")
print(f"内容: {ai_msg.content}")
print("作用: 模型的回复，可能包含工具调用")
print()

print("="*60)
print("4. ToolMessage - 工具结果")
print("="*60)
tool_msg = ToolMessage(
    content="搜索结果: 找到3篇相关文档",
    tool_call_id="call_123"
)
print(f"类型: {tool_msg.type}")
print(f"内容: {tool_msg.content}")
print(f"工具调用ID: {tool_msg.tool_call_id}")
print("作用: 将工具执行结果返回给模型")
print()

print("💡 小白提示：")
print("消息类型的顺序通常是：System → Human → AI → Tool → AI → ...")
````

### 3.4 实战：构建多轮对话

理解消息序列对于构建对话系统至关重要：

````python
# 构建一个数学辅导对话
conversation = [
    SystemMessage(content="你是一个耐心的数学老师，用简单的例子解释概念"),
]

# 第一轮对话
print("用户: 什么是质数？")
conversation.append(HumanMessage(content="什么是质数？"))

response1 = deepseek_model.invoke(conversation)
conversation.append(response1)
print(f"AI: {response1.content}\n")

# 第二轮对话（AI会记住之前的上下文）
print("用户: 给我举3个例子")
conversation.append(HumanMessage(content="给我举3个例子"))
# print(conversation)
response2 = deepseek_model.invoke(conversation)
conversation.append(response2)
print(f"AI: {response2.content}\n")

# 第三轮对话
print("用户: 为什么1不是质数？")
conversation.append(HumanMessage(content="为什么1不是质数？"))

response3 = deepseek_model.invoke(conversation)
print(f"AI: {response3.content}\n")

print("="*60)
print(f"💡 对话共{len(conversation)}条消息")
print("消息类型序列:", " → ".join([msg.type for msg in conversation]))
````

## 第五章：Tools（工具）- 扩展模型的能力

### 5.1 什么是Tools？

**Tools（工具）**是LangChain中最强大的概念之一。模型本身只能生成文本，但通过工具，模型可以：
- 📊 查询数据库
- 🌐 访问互联网
- 🧮 执行计算
- 📁 读写文件
- 🔌 调用API

#### 核心概念

```
┌─────────────────────────────────────────┐
│           用户问题                       │
│   "北京今天天气如何？"                    │
└──────────────┬──────────────────────────┘
               │
               ▼
┌─────────────────────────────────────────┐
│           LLM 分析                       │
│   "需要调用天气查询工具"                  │
└──────────────┬──────────────────────────┘
               │
               ▼
┌─────────────────────────────────────────┐
│         工具执行                         │
│   get_weather(city="北京")              │
│   → "北京：晴，22°C"                     │
└──────────────┬──────────────────────────┘
               │
               ▼
┌─────────────────────────────────────────┐
│         LLM 生成回答                     │
│   "北京今天天气晴朗，温度22°C"            │
└─────────────────────────────────────────┘
```

### 5.2 创建第一个工具

在LangChain中，创建工具有三种方式：

````python
from langchain.tools import tool
from langchain_core.tools import StructuredTool

print("="*60)
print("方式1：使用@tool装饰器（推荐，最简单）")
print("="*60)

@tool
def get_weather(location: str) -> str:
    """获取指定位置的当前天气信息。
    
    Args:
        location: 城市名称，如"北京"、"上海"
    """
    # 模拟天气查询
    weather_data = {
        "北京": "晴天，温度22°C",
        "上海": "多云，温度25°C",
        "深圳": "小雨，温度28°C"
    }
    return weather_data.get(location, f"{location}的天气：晴朗，20°C")

print("工具名称:", get_weather.name)
print("工具描述:", get_weather.description)
print("测试调用:", get_weather.invoke({"location": "北京"}))
print()

print("="*60)
print("方式2：使用@tool装饰器 + 详细参数说明")
print("="*60)

@tool
def calculator(
    operation: str,
    num1: float,
    num2: float
) -> str:
    """执行基本数学运算。
    
    Args:
        operation: 运算类型，可选值："add"(加), "subtract"(减), "multiply"(乘), "divide"(除)
        num1: 第一个数字
        num2: 第二个数字
    
    Returns:
        str: 计算结果的字符串表示
    """
    ops = {
        "add": lambda a, b: a + b,
        "subtract": lambda a, b: a - b,
        "multiply": lambda a, b: a * b,
        "divide": lambda a, b: a / b if b != 0 else "错误：除数不能为0"
    }
    
    if operation not in ops:
        return f"不支持的运算：{operation}"
    
    result = ops[operation](num1, num2)
    return f"{num1} {operation} {num2} = {result}"

print("工具名称:", calculator.name)
print("工具参数:", calculator.args)
print("测试调用:", calculator.invoke({"operation": "multiply", "num1": 5, "num2": 7}))
print()

print("="*60)
print("方式3：使用StructuredTool（最灵活）")
print("="*60)

def search_database_func(query: str, limit: int = 5):
    """实际的搜索逻辑"""
    return f"在数据库中找到 {limit} 条关于'{query}'的结果"

search_database = StructuredTool.from_function(
    func=search_database_func,
    name="search_database",
    description="在客户数据库中搜索记录，支持关键词匹配",
)

print("工具名称:", search_database.name)
print("测试调用:", search_database.invoke({"query": "Python", "limit": 3}))
print()

print("💡 小白提示：")
print("- 使用@tool最简单，适合90%的场景")
print("- 工具描述要清晰，模型靠这个判断何时调用")
print("- 参数类型提示很重要，帮助模型正确传参")
````

### 4.4 工具绑定 - 让模型知道有哪些工具可用

绑定工具后，模型可以"看到"工具列表，并决定是否需要调用：

````python
# 将工具绑定到模型
tools = [get_weather, calculator]
model_with_tools = deepseek_model.bind_tools(tools)

print("="*60)
print("测试1：模型决定调用工具")
print("="*60)
response = model_with_tools.invoke("北京今天天气怎么样？")

# 检查response对象是否包含 tool_calls 属性 且tool_calls 属性不是空的
if hasattr(response, 'tool_calls') and response.tool_calls:
    print("✅ 模型决定调用工具:")
    for tool_call in response.tool_calls:
        print(f"  工具名: {tool_call['name']}")
        print(f"  参数: {tool_call['args']}")
        print(f"  调用ID: {tool_call['id']}")
else:
    print("❌ 模型没有调用工具")
    print("响应:", response.content)
print()

print("="*60)
print("测试2：模型决定不调用工具")
print("="*60)
response2 = model_with_tools.invoke("你好，今天心情怎么样？")

if hasattr(response2, 'tool_calls') and response2.tool_calls:
    print("✅ 模型调用了工具")
else:
    print("❌ 模型没有调用工具（因为不需要）")
    print("响应:", response2.content)
print()

print("="*60)
print("测试3：模型可能同时调用多个工具")
print("="*60)
response3 = model_with_tools.invoke("北京的天气如何？另外帮我算一下25乘以4")

if hasattr(response3, 'tool_calls') and response3.tool_calls:
    print(f"✅ 模型调用了 {len(response3.tool_calls)} 个工具:")
    for i, tool_call in enumerate(response3.tool_calls, 1):
        print(f"\n  工具{i}: {tool_call['name']}")
        print(f"  参数: {tool_call['args']}")
print()

print("💡 关键点：")
print("1. bind_tools() 后，模型获得了工具的'能力'")
print("2. 模型会根据用户问题智能决定是否调用工具")
print("3. tool_calls 包含了模型想要调用的工具信息")
print("4. 我们还需要手动执行工具并返回结果（下一节）")
````

### 4.5 工具调用循环 - 完整的执行流程

这是理解工具调用的核心！流程如下：

```
用户问题 → 模型分析 → 决定调用工具 → 执行工具 → 返回结果 → 模型生成最终答案
```

````python
from langchain.messages import ToolMessage

def run_tool_loop_v2(user_input: str, verbose=True):
    """完整的工具调用循环，带详细日志"""
    
    if verbose:
        print(f"👤 用户: {user_input}")
        print("="*60)
    
    # 1. 构建消息列表
    messages = [HumanMessage(content=user_input)]
    
    # 2. 第一次调用：让模型决定是否需要工具
    if verbose:
        print("🤔 模型思考中...")
    response = model_with_tools.invoke(messages)
    messages.append(response)
    
    # 3. 检查是否有工具调用
    if not hasattr(response, 'tool_calls') or not response.tool_calls:
        if verbose:
            print("✅ 无需工具，直接回答")
        return response.content
    
    # 4. 执行所有工具调用
    if verbose:
        print(f"🔧 需要调用 {len(response.tool_calls)} 个工具:\n")
    
    # 创建工具字典以便查找
    tool_map = {tool.name: tool for tool in tools}
    
    for i, tool_call in enumerate(response.tool_calls, 1):
        tool_name = tool_call["name"]
        tool_args = tool_call["args"]
        tool_id = tool_call["id"]
        
        if verbose:
            print(f"  [{i}] 工具: {tool_name}")
            print(f"      参数: {tool_args}")
        
        # 执行工具
        if tool_name in tool_map:
            result = tool_map[tool_name].invoke(tool_args)
            if verbose:
                print(f"      结果: {result}\n")
        else:
            result = f"错误: 未找到工具 {tool_name}"
            if verbose:
                print(f"      {result}\n")
        
        # 将工具结果添加到消息列表
        messages.append(ToolMessage(
            content=result,
            tool_call_id=tool_id
        ))
    
    # 5. 第二次调用：基于工具结果生成最终答案
    if verbose:
        print("🤔 模型整合结果中...")
    final_response = model_with_tools.invoke(messages)
    
    if verbose:
        print("="*60)
        print(f"🤖 最终回答: {final_response.content}")
        print("="*60)
        print(f"\n📊 消息统计: 共 {len(messages) + 1} 条消息")
        print(f"   消息流: {' → '.join([msg.type for msg in messages])} → ai")
    
    return final_response.content

# 测试不同场景
print("\n" + "🌟"*30)
print("场景1: 单工具调用")
print("🌟"*30)
run_tool_loop_v2("深圳今天天气如何？")

print("\n\n" + "🌟"*30)
print("场景2: 多工具调用")
print("🌟"*30)
run_tool_loop_v2("北京天气怎样？还有，12乘以8等于多少？")

print("\n\n" + "🌟"*30)
print("场景3: 无需工具")
print("🌟"*30)
run_tool_loop_v2("给我讲个笑话")
````

## 第六章：Agents（代理）- 自主决策系统

### 6.1 什么是Agent？深入理解

**Agent（代理）**是LangChain v1.0的核心功能,它不仅仅是一个工具调用器，而是一个能够：

- 🧠 **自主决策**：根据任务自行决定使用哪些工具
- 🔄 **多步推理**：将复杂任务分解为多个子任务
- 🎯 **目标导向**：持续工作直到完成目标
- 🛠️ **工具编排**：智能地组合和使用多个工具

#### Agent vs 简单工具调用

```
简单工具调用:
用户问题 → 模型 → 调用1个工具 → 返回结果

Agent (智能代理):
用户问题 → Agent思考 → 调用工具A → 评估结果 → 调用工具B → 再思考 → ... → 最终答案
```

### 6.2 Agent的核心组件

1. **大语言模型 (LLM)**: Agent的"大脑"
2. **工具集 (Tools)**: Agent可以使用的能力
3. **提示词 (Prompt)**: 指导Agent的行为
4. **记忆 (Memory)**: 记住对话历史（可选）
5. **输出解析器 (Parser)**: 理解Agent的决策

## 第七章：Streaming（流式输出）- 实时响应体验

### 6.1 为什么需要流式输出？

在实际应用中，LLM生成响应可能需要几秒甚至更长时间。流式输出可以：

- ⚡ **提升用户体验**：立即看到响应开始生成
- 🎯 **降低等待焦虑**：用户知道系统在工作
- 🚀 **更快的感知速度**：即使总时间相同，感觉更快
- 💡 **实时交互**：可以提前中断或调整

### 6.2 流式输出的三种层级

```
┌──────────────────────────────────────────┐
│ Level 1: 模型流式输出                      │
│ - 逐token生成文本                          │
│ - 最基础的流式功能                         │
└──────────────────────────────────────────┘
           ↓
┌──────────────────────────────────────────┐
│ Level 2: Chain流式输出                    │
│ - 链中每个步骤的输出                       │
│ - 可以看到中间过程                         │
└──────────────────────────────────────────┘
           ↓
┌──────────────────────────────────────────┐
│ Level 3: Agent流式输出                    │
│ - Agent的思考过程                          │
│ - 工具调用和结果                           │
│ - 最完整的可观察性                         │
└──────────────────────────────────────────┘
```

### 6.3 基础模型流式输出

````python
from langchain.agents import create_react_agent
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# 定义工具集
@tool
def search(query: str) -> str:
    """在互联网上搜索信息。当需要查找最新信息或不确定的事实时使用。"""
    # 模拟搜索
    results = {
        "Python最佳实践": "1. 使用虚拟环境 2. 遵循PEP8规范 3. 编写测试",
        "LangChain教程": "LangChain是一个强大的LLM应用框架...",
    }
    for key in results:
        if key.lower() in query.lower():
            return f"搜索'{query}'的结果：{results[key]}"
    return f"搜索'{query}'的结果：找到了相关信息。"

@tool
def calculator(expression: str) -> str:
    """计算数学表达式。当需要进行精确计算时使用。
    
    Args:
        expression: 数学表达式，如 "2+3*4" 或 "sqrt(16)"
    """
    try:
        # 安全的数学计算
        import math
        allowed_names = {
            'sqrt': math.sqrt,
            'pow': math.pow,
            'sin': math.sin,
            'cos': math.cos,
        }
        result = eval(expression, {"__builtins__": None}, allowed_names)
        return f"{expression} = {result}"
    except Exception as e:
        return f"计算错误：{str(e)}"

@tool
def get_current_time() -> str:
    """获取当前日期和时间。"""
    from datetime import datetime
    return datetime.now().strftime("%Y年%m月%d日 %H:%M:%S")

# 创建Agent
print("="*60)
print("创建Agent的关键参数：")
print("="*60)

agent = create_react_agent(
    llm=openai_model,
    tools=[search, calculator, get_current_time],
    prompt=ChatPromptTemplate.from_messages([
        ("system", """你是一个有用的AI助手。你可以使用以下工具来帮助用户：

- search: 搜索信息
- calculator: 进行数学计算
- get_current_time: 获取当前时间

思考过程：
1. 理解用户的问题
2. 决定需要使用哪些工具
3. 调用工具获取信息
4. 基于工具结果给出最终答案

重要：如果可以直接回答，就不要使用工具。只在真正需要时才使用工具。"""),
        MessagesPlaceholder(variable_name="chat_history", optional=True),
        ("user", "{input}"),
        MessagesPlaceholder(variable_name="agent_scratchpad"),
    ]),
)

print("✅ Agent创建成功！")
print("\n💡 Agent包含：")
print("  - 模型：GPT-4o")
print("  - 工具：3个（搜索、计算器、时间）")
print("  - 系统提示：定义Agent的行为和能力")
````

````python
import time

print("="*60)
print("演示1：基础流式输出 - 逐token显示")
print("="*60)

prompt = "请写一首关于人工智能的五言绝句"
print(f"提示：{prompt}\n")
print("AI响应：", end="", flush=True)

for chunk in deepseek_model.stream(prompt):
    print(chunk.content, end="", flush=True)
    time.sleep(0.05)  # 模拟打字效果

print("\n\n" + "="*60)
print("演示2：获取完整的流式元数据")
print("="*60)

total_tokens = 0
chunk_count = 0

for chunk in deepseek_model.stream("简单介绍一下Python"):
    chunk_count += 1
    print(f"Chunk {chunk_count}: {chunk.content}", end="", flush=True)
    
    # 某些chunk可能包含元数据
    if hasattr(chunk, 'response_metadata'):
        metadata = chunk.response_metadata
        if 'token_usage' in metadata:
            total_tokens = metadata['token_usage'].get('total_tokens', 0)

print(f"\n\n📊 统计：共 {chunk_count} 个chunks")
if total_tokens > 0:
    print(f"总tokens: {total_tokens}")

print("\n💡 小白提示：")
print("- 每个chunk是生成的一小段文本")
print("- flush=True确保立即显示，不缓冲")
print("- 流式输出特别适合长文本生成")
````

### 6.4 Chain流式输出 - 观察处理流程

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 创建一个简单的Chain
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个专业的{role}。"),
    ("user", "{question}")
])

# 这里使用了 LCEL (LangChain Expression Language) 语法 | 将三个组件串联起来：
# prompt：负责把用户的输入（如 "Python的优势"）包装成完整的提示词。
# deepseek_model：负责接收提示词并生成回答。
# StrOutputParser()：负责把模型返回的复杂对象（AIMessage）直接转成纯字符串，方便后续处理。
chain = prompt | deepseek_model | StrOutputParser()

print("="*60)
print("Chain流式输出示例")
print("="*60)
print("\n问题：Python的优势是什么？\n")
print("回答：", end="", flush=True)

# 流式执行Chain
for chunk in chain.stream({
    "role": "Python开发专家",
    "question": "请用三点总结Python的主要优势"
}):
    print(chunk, end="", flush=True)

print("\n\n💡 Chain流式的优势：")
print("- 可以实时看到每个组件的输出")
print("- 适合多步骤处理的可视化")
print("- 用户体验更好")
````

````python
from pydantic import BaseModel, Field
from typing import List, Optional

# 定义输出结构
class Person(BaseModel):
    """人员信息"""
    name: str = Field(description="人的姓名")
    age: int = Field(description="年龄，必须是整数")
    city: str = Field(description="居住城市")
    email: Optional[str] = Field(default=None, description="电子邮箱（可选）")

class MovieReview(BaseModel):
    """电影评论"""
    title: str = Field(description="电影名称")
    rating: float = Field(description="评分，1-10之间", ge=1, le=10)
    pros: List[str] = Field(description="优点列表")
    cons: List[str] = Field(description="缺点列表")
    recommend: bool = Field(description="是否推荐")

print("="*60)
print("示例1：提取人员信息")
print("="*60)

# 使用with_structured_output方法
structured_model = openai_model.with_structured_output(Person)

text = "张三今年28岁，住在北京，邮箱是zhangsan@email.com"
result = structured_model.invoke(f"从以下文本提取信息：{text}")

print(f"输入文本：{text}")
print(f"\n提取结果（Pydantic对象）：")
print(f"  类型: {type(result)}")
print(f"  姓名: {result.name}")
print(f"  年龄: {result.age} (类型: {type(result.age).__name__})")
print(f"  城市: {result.city}")
print(f"  邮箱: {result.email}")

# 转换为字典
print(f"\n转换为字典：{result.dict()}")

# 转换为JSON
import json
print(f"\n转换为JSON：{json.dumps(result.dict(), ensure_ascii=False)}")

print("\n" + "="*60)
print("示例2：生成电影评论")
print("="*60)

review_model = openai_model.with_structured_output(MovieReview)

result = review_model.invoke("请评论电影《肖申克的救赎》")

print(f"电影：{result.title}")
print(f"评分：{result.rating}")
print(f"优点：{', '.join(result.pros)}")
print(f"缺点：{', '.join(result.cons)}")
print(f"推荐：{'是' if result.recommend else '否'}")

print("\n💡 结构化输出的优势：")
print("1. 类型安全：age一定是int，不会是字符串")
print("2. 验证自动：rating自动验证范围1-10")
print("3. 易于使用：可以直接访问属性result.name")
print("4. 文档清晰：Field描述说明每个字段的含义")
````

### 7.3 复杂嵌套结构

Pydantic支持复杂的嵌套结构，适合真实业务场景：

````python
from datetime import datetime
from enum import Enum

class OrderStatus(str, Enum):
    """订单状态枚举"""
    PENDING = "pending"
    PAID = "paid"
    SHIPPED = "shipped"
    DELIVERED = "delivered"

class OrderItem(BaseModel):
    """订单商品"""
    product_name: str = Field(description="商品名称")
    quantity: int = Field(description="数量", gt=0)
    price: float = Field(description="单价", gt=0)

class Order(BaseModel):
    """订单信息"""
    order_id: str = Field(description="订单编号")
    customer_name: str = Field(description="客户姓名")
    items: List[OrderItem] = Field(description="商品列表")
    status: OrderStatus = Field(description="订单状态")
    total_amount: float = Field(description="总金额")
    notes: Optional[str] = Field(default=None, description="备注")

# 创建结构化模型
order_model = openai_model.with_structured_output(Order)

# 测试
print("="*60)
print("从自然语言生成结构化订单")
print("="*60)

text = """
客户李四下了一个订单，订单号是ORD20241125001。
他购买了2个笔记本电脑，每个5999元，还有1个鼠标99元。
订单已经支付，总共12097元。备注：请尽快发货。
"""

result = order_model.invoke(f"解析以下订单信息：{text}")

print(f"订单编号：{result.order_id}")
print(f"客户：{result.customer_name}")
print(f"状态：{result.status.value}")
print(f"\n商品清单：")
for i, item in enumerate(result.items, 1):
    print(f"  {i}. {item.product_name} x{item.quantity} = ¥{item.price * item.quantity}")
print(f"\n总金额：¥{result.total_amount}")
print(f"备注：{result.notes}")

# 验证数据
print(f"\n✅ 数据验证：")
print(f"  - items是列表：{isinstance(result.items, list)}")
print(f"  - 第一个item是OrderItem对象：{isinstance(result.items[0], OrderItem)}")
print(f"  - status是枚举类型：{isinstance(result.status, OrderStatus)}")

print("\n💡 嵌套结构的优势：")
print("- 可以表达复杂的业务数据")
print("- 类型安全在所有层级都有保障")
print("- 使用枚举限定可选值")
print("- 可以直接用于数据库存储或API响应")
````

## 第九章：Streaming（流式输出）- 实时响应体验

### 7.1 为什么需要Streaming？

在生成长文本时，如果等待完整响应会导致：
- ⏰ 用户等待时间过长
- 😴 界面卡顿，体验差
- 🔄 无法实时看到进度

**Streaming（流式输出）**解决了这些问题：

```
传统方式:
请求 → [等待5秒...] → 完整响应显示

流式方式:
请求 → 显示"今" → 显示"天" → 显示"天气" → ...
```

### 7.2 基础流式输出

````python
from langchain.schema import HumanMessage, AIMessage, SystemMessage

print("="*60)
print("演示：手动管理对话历史")
print("="*60)

# 初始化对话历史
chat_history = [
    SystemMessage(content="你是一个友好的助手，擅长记住用户信息。")
]

def chat(user_input: str):
    """模拟对话函数"""
    # 添加用户消息
    chat_history.append(HumanMessage(content=user_input))
    
    # 调用模型（带上完整历史）
    response = openai_model.invoke(chat_history)
    
    # 添加AI响应到历史
    chat_history.append(response)
    
    return response.content

# 多轮对话测试
print("\n👤 用户: 我叫张三，今年28岁")
print(f"🤖 AI: {chat('我叫张三，今年28岁')}")

print("\n👤 用户: 我喜欢打篮球")
print(f"🤖 AI: {chat('我喜欢打篮球')}")

print("\n👤 用户: 我的名字是什么？")
print(f"🤖 AI: {chat('我的名字是什么？')}")

print("\n👤 用户: 我多大了？")
print(f"🤖 AI: {chat('我多大了？')}")

print("\n👤 用户: 我的爱好是什么？")
print(f"🤖 AI: {chat('我的爱好是什么？')}")

# 查看对话历史
print("\n" + "="*60)
print(f"📊 对话统计：共 {len(chat_history)} 条消息")
print("="*60)
for i, msg in enumerate(chat_history, 1):
    role = msg.type
    content = msg.content[:50] + "..." if len(msg.content) > 50 else msg.content
    print(f"{i}. [{role}] {content}")

print("\n💡 手动管理的问题：")
print("- 需要手动维护chat_history列表")
print("- 历史会无限增长，消耗token")
print("- 需要自己处理窗口截断")
````

### 8.2 使用Pydantic定义输出结构

````python
import time
from functools import wraps

print("="*60)
print("中间件示例1: 日志记录中间件")
print("="*60)

class LoggingMiddleware:
    """日志记录中间件"""
    
    def __init__(self, llm):
        self.llm = llm
        self.call_count = 0
    
    def invoke(self, input_data):
        self.call_count += 1
        
        print(f"\n📝 [日志 #{self.call_count}]")
        print(f"⏰ 时间: {time.strftime('%H:%M:%S')}")
        print(f"📥 输入: {input_data[:100]}...")
        
        # 调用实际的LLM
        start_time = time.time()
        result = self.llm.invoke(input_data)
        elapsed = time.time() - start_time
        
        print(f"⏱️  耗时: {elapsed:.2f}秒")
        print(f"📤 输出: {result.content[:100]}...")
        
        return result

# 使用日志中间件
logging_model = LoggingMiddleware(openai_model)

response = logging_model.invoke("什么是Python？")
response = logging_model.invoke("什么是机器学习？")

print(f"\n📊 总调用次数: {logging_model.call_count}")

print("\n\n" + "="*60)
print("中间件示例2: 错误重试中间件")
print("="*60)

class RetryMiddleware:
    """自动重试中间件"""
    
    def __init__(self, llm, max_retries=3):
        self.llm = llm
        self.max_retries = max_retries
    
    def invoke(self, input_data):
        for attempt in range(self.max_retries):
            try:
                print(f"🔄 尝试 {attempt + 1}/{self.max_retries}")
                result = self.llm.invoke(input_data)
                print(f"✅ 成功")
                return result
            except Exception as e:
                print(f"❌ 失败: {str(e)}")
                if attempt == self.max_retries - 1:
                    raise
                time.sleep(1)  # 等待后重试

# 使用重试中间件
retry_model = RetryMiddleware(openai_model, max_retries=3)

try:
    response = retry_model.invoke("你好")
except Exception as e:
    print(f"最终失败: {e}")

print("\n\n" + "="*60)
print("中间件示例3: 敏感信息过滤中间件")
print("="*60)

class SensitiveDataFilter:
    """敏感数据过滤中间件"""
    
    def __init__(self, llm):
        self.llm = llm
        self.sensitive_patterns = [
            (r'\d{3}-\d{4}-\d{4}', '[电话已隐藏]'),  # 电话号码
            (r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', '[邮箱已隐藏]'),  # 邮箱
            (r'\d{15,18}', '[身份证已隐藏]'),  # 身份证
        ]
    
    def filter_sensitive(self, text):
        """过滤敏感信息"""
        import re
        filtered = text
        for pattern, replacement in self.sensitive_patterns:
            filtered = re.sub(pattern, replacement, filtered)
        return filtered
    
    def invoke(self, input_data):
        # 过滤输入
        filtered_input = self.filter_sensitive(input_data)
        print(f"原始输入: {input_data}")
        print(f"过滤后: {filtered_input}")
        
        # 调用LLM
        result = self.llm.invoke(filtered_input)
        
        # 也可以过滤输出（如果需要）
        return result

# 使用过滤中间件
filter_model = SensitiveDataFilter(openai_model)

response = filter_model.invoke(
    "我的邮箱是john@example.com，电话是123-4567-8901"
)

print("\n💡 中间件的应用场景：")
print("1. 日志记录：审计、调试、分析")
print("2. 错误处理：自动重试、降级策略")
print("3. 安全过滤：敏感信息保护")
print("4. 性能监控：响应时间、成功率")
print("5. 缓存：减少重复调用")
print("6. 限流：控制API调用频率")
````

## 第十一章：Short-term Memory（短期记忆）- 对话上下文管理

### 9.1 为什么需要Memory？

LLM本身是**无状态**的，每次调用都是独立的。但实际对话需要：

```
用户: 我叫张三
AI: 你好张三！

用户: 我叫什么？
AI: (如果没有记忆) 我不知道你的名字
    (如果有记忆) 你叫张三
```

**Memory（记忆）**解决的问题：
- 💬 维护对话上下文
- 🧠 记住之前的信息
- 🔄 保持对话连贯性

### 9.2 Memory的核心概念

````python
class MiddlewareChain:
    """中间件链，支持组合多个中间件"""
    
    def __init__(self, llm, middlewares=None):
        self.llm = llm
        self.middlewares = middlewares or []
    
    def add_middleware(self, middleware):
        """添加中间件"""
        self.middlewares.append(middleware)
    
    def invoke(self, input_data):
        """执行带所有中间件的调用"""
        # 前置处理
        processed_input = input_data
        for middleware in self.middlewares:
            if hasattr(middleware, 'before_invoke'):
                processed_input = middleware.before_invoke(processed_input)
        
        # 实际调用
        result = self.llm.invoke(processed_input)
        
        # 后置处理
        processed_result = result
        for middleware in reversed(self.middlewares):
            if hasattr(middleware, 'after_invoke'):
                processed_result = middleware.after_invoke(processed_result)
        
        return processed_result

# 定义几个简单中间件
class TimingMiddleware:
    """计时中间件"""
    def before_invoke(self, input_data):
        self.start_time = time.time()
        print("⏱️  开始计时...")
        return input_data
    
    def after_invoke(self, result):
        elapsed = time.time() - self.start_time
        print(f"⏱️  耗时: {elapsed:.2f}秒")
        return result

class UpperCaseMiddleware:
    """转大写中间件（示例）"""
    def before_invoke(self, input_data):
        print("🔤 输入转大写")
        return input_data.upper()
    
    def after_invoke(self, result):
        return result

class CounterMiddleware:
    """计数中间件"""
    def __init__(self):
        self.count = 0
    
    def before_invoke(self, input_data):
        self.count += 1
        print(f"📊 第 {self.count} 次调用")
        return input_data
    
    def after_invoke(self, result):
        return result

# 创建中间件链
print("="*60)
print("组合多个中间件")
print("="*60)

chain = MiddlewareChain(openai_model)
chain.add_middleware(CounterMiddleware())
chain.add_middleware(TimingMiddleware())

# 测试
print("\n调用1:")
result = chain.invoke("hello")

print("\n调用2:")
result = chain.invoke("world")

print("\n💡 中间件链的优势：")
print("- 模块化：每个中间件负责一个功能")
print("- 可组合：灵活组合不同中间件")
print("- 可重用：中间件可在不同地方使用")
print("- 易维护：修改某个功能只需改对应中间件")
````

## 第十二章：RAG（检索增强生成）- 知识库集成

### 10.1 什么是RAG？

**RAG（Retrieval Augmented Generation）**是一种强大的技术，它结合了：
- 🔍 **检索（Retrieval）**：从知识库中找到相关信息
- 🤖 **生成（Generation）**：基于检索结果生成答案

### 10.2 为什么需要RAG？

LLM虽然强大，但有局限：
- ❌ 训练数据有时效性（不知道最新信息）
- ❌ 不了解企业私有数据
- ❌ 可能产生幻觉（编造信息）

RAG的优势：
- ✅ 访问最新和私有数据
- ✅ 减少幻觉，答案有依据
- ✅ 可追溯信息来源
- ✅ 无需重新训练模型

### 10.3 RAG工作流程

```
用户问题
    ↓
向量化查询
    ↓
向量数据库检索 → 找到相关文档
    ↓
构建提示（问题 + 文档）
    ↓
LLM生成答案
    ↓
返回结果（附带来源）
```

````python
from langchain_community.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings

print("="*60)
print("步骤1: 准备文档数据")
print("="*60)

# 示例文档 - 模拟企业知识库
documents = [
    "LangChain v1.0是一个专注于生产环境的Agent构建框架，引入了create_agent函数。",
    "RAG技术结合了检索和生成，可以让LLM访问外部知识库，提供更准确的答案。",
    "工具调用（Tool Calling）允许LLM与外部系统交互，执行实际操作。",
    "LangGraph是LangChain的底层编排框架，支持复杂的Agent工作流。",
    "Streaming流式输出可以实时显示LLM生成的内容，提升用户体验。",
    "Pydantic模型可以定义结构化输出，确保LLM返回格式化的数据。",
    "Memory组件帮助管理对话历史，让Agent能够记住之前的交互。",
    "Middleware中间件提供了在请求响应流程中插入自定义逻辑的能力。",
    "向量数据库如FAISS、Chroma、Pinecone用于存储和检索文档嵌入。",
    "Embedding嵌入模型将文本转换为向量，使语义搜索成为可能。"
]

print(f"📚 文档数量: {len(documents)}")
print(f"📄 示例文档: {documents[0][:50]}...")

print("\n" + "="*60)
print("步骤2: 分割文档")
print("="*60)

# 文档分割器
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,      # 每个块的大小
    chunk_overlap=20,    # 块之间的重叠
    length_function=len,
    separators=["\n\n", "\n", "。", "，", " ", ""]
)

splits = text_splitter.create_documents(documents)
print(f"✂️  分割为 {len(splits)} 个片段")
print(f"📄 第一个片段: {splits[0].page_content}")

print("\n💡 为什么要分割文档？")
print("- LLM有token限制，不能一次处理太多文本")
print("- 小块更精确，检索更准确")
print("- 重叠确保不会丢失跨边界的信息")
````

### 10.5 创建向量存储和检索器

````python
print("="*60)
print("步骤3: 创建Embedding模型")
print("="*60)

# 创建嵌入模型（将文本转换为向量）
embeddings = OpenAIEmbeddings()

# 测试嵌入
sample_text = "什么是LangChain？"
sample_vector = embeddings.embed_query(sample_text)
print(f"📝 原文: {sample_text}")
print(f"🔢 向量维度: {len(sample_vector)}")
print(f"🔢 向量示例: [{sample_vector[0]:.4f}, {sample_vector[1]:.4f}, ...]")

print("\n" + "="*60)
print("步骤4: 创建向量数据库")
print("="*60)

# 创建FAISS向量存储
vector_store = FAISS.from_documents(splits, embeddings)
print("✅ 向量数据库创建成功")
print(f"📊 存储了 {len(splits)} 个文档片段")

print("\n" + "="*60)
print("步骤5: 测试相似度搜索")
print("="*60)

# 测试检索
query = "什么是RAG？"
print(f"🔍 查询: {query}\n")

# 相似度搜索
results = vector_store.similarity_search(query, k=3)

for i, doc in enumerate(results, 1):
    print(f"结果{i}: {doc.page_content}")
    print()

print("="*60)
print("步骤6: 创建检索器")
print("="*60)

# 创建检索器（Retriever）
retriever = vector_store.as_retriever(
    search_type="similarity",    # 搜索类型：similarity或mmr
    search_kwargs={"k": 2}       # 返回最相关的2个结果
)

print("✅ 检索器创建成功")
print("配置: 返回最相关的2个文档")

# 测试检索器
docs = retriever.invoke("什么是Streaming？")
print(f"\n检索到 {len(docs)} 个相关文档:")
for i, doc in enumerate(docs, 1):
    print(f"{i}. {doc.page_content}")
````

## 第十三章：Middleware（中间件）- 增强与定制

### 10.1 什么是Middleware？

**Middleware（中间件）**是一种设计模式，可以在请求/响应流程中插入自定义逻辑：

```
用户请求
    ↓
中间件1 (日志记录)
    ↓
中间件2 (权限验证)
    ↓
核心处理 (LLM调用)
    ↓
中间件2 (性能统计)
    ↓
中间件1 (结果缓存)
    ↓
返回响应
```

**使用场景：**
- 📊 监控和日志
- 🔐 权限控制
- ⚡ 性能优化
- 🎯 内容过滤
- 💾 缓存策略

### 10.2 实现基础中间件

````python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

print("="*60)
print("方式1: 简单RAG链（适合单次问答）")
print("="*60)

# 创建RAG提示模板
rag_prompt = ChatPromptTemplate.from_messages([
    ("system", """你是一个AI助手，请基于以下上下文回答问题。

上下文：
{context}

如果上下文中没有相关信息，请诚实地说"我不知道"。
不要编造信息。"""),
    ("user", "{question}")
])

# 格式化文档函数
def format_docs(docs):
    """将检索到的文档格式化为字符串"""
    return "\n\n".join(f"文档{i+1}: {doc.page_content}" for i, doc in enumerate(docs))

# 创建RAG链
rag_chain = (
    {
        "context": retriever | format_docs,  # 检索并格式化
        "question": RunnablePassthrough()     # 直接传递问题
    }
    | rag_prompt          # 构建提示
    | openai_model        # 调用模型
    | StrOutputParser()   # 解析输出
)

# 测试问题
questions = [
    "什么是RAG技术？",
    "LangChain v1.0有什么新特性？",
    "如何使用Pydantic？",
    "什么是区块链？"  # 知识库中没有的问题
]

for q in questions:
    print(f"\n❓ 问题: {q}")
    answer = rag_chain.invoke(q)
    print(f"💡 答案: {answer}")
    print("-" * 60)

print("\n💡 简单RAG链的特点：")
print("✅ 实现简单，代码清晰")
print("✅ 适合单轮问答")
print("❌ 不支持多轮对话")
print("❌ 不能根据答案进行追问")
````

## 第十四章：RAG（检索增强生成）- 知识库集成

### 11.1 什么是RAG？

**RAG (Retrieval Augmented Generation)** 解决了LLM的两大核心问题：

1. **知识过时**：模型训练数据有截止日期
2. **幻觉问题**：模型可能编造不存在的信息

#### RAG工作原理

```
┌─────────────────────────────────────────────────┐
│  步骤1: 构建知识库                               │
│  文档 → 分块 → 向量化 → 存储到向量数据库          │
└─────────────────────────────────────────────────┘
                    ↓
┌─────────────────────────────────────────────────┐
│  步骤2: 用户查询                                 │
│  "LangChain是什么?" → 向量化                     │
└─────────────────────────────────────────────────┘
                    ↓
┌─────────────────────────────────────────────────┐
│  步骤3: 检索相关文档                             │
│  在向量库中找到最相似的文档片段                   │
└─────────────────────────────────────────────────┘
                    ↓
┌─────────────────────────────────────────────────┐
│  步骤4: 生成答案                                 │
│  LLM基于检索到的文档 + 用户问题 → 生成答案        │
└─────────────────────────────────────────────────┘
```

### 11.2 RAG的核心组件

### 7.4 综合案例：智能客服系统（整合所有核心组件）

这个案例整合了我们学到的所有核心组件：

````python
from pydantic import BaseModel, Field
from typing import List, Optional
from langchain.memory import ConversationBufferMemory
from langchain.agents import AgentExecutor, create_react_agent
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

print("="*60)
print("🎯 综合案例：智能电商客服系统")
print("="*60)
print("\n整合组件：")
print("✅ Models - GPT-4o")
print("✅ Messages - 对话消息管理")
print("✅ Tools - 订单查询、库存检查等")
print("✅ Agent - 自主决策")
print("✅ Structured Output - 结构化返回")
print("✅ Memory - 记住用户信息")
print("✅ Streaming - 实时响应（可选）")

# 1. 定义结构化输出
class CustomerInfo(BaseModel):
    """客户信息"""
    name: str = Field(description="客户姓名")
    member_level: str = Field(description="会员等级：普通/银卡/金卡/钻石")
    phone: Optional[str] = Field(default=None, description="电话号码")

class OrderInfo(BaseModel):
    """订单信息"""
    order_id: str = Field(description="订单编号")
    status: str = Field(description="订单状态")
    items: List[str] = Field(description="商品列表")
    total: float = Field(description="总金额")

# 2. 定义工具集
@tool
def query_order(order_id: str) -> str:
    """查询订单状态和详情"""
    orders = {
        "ORD001": OrderInfo(
            order_id="ORD001",
            status="已发货",
            items=["iPhone 15", "AirPods"],
            total=8999.0
        ),
        "ORD002": OrderInfo(
            order_id="ORD002",
            status="待支付",
            items=["MacBook Pro"],
            total=15999.0
        ),
    }
    order = orders.get(order_id)
    if order:
        return f"订单{order_id}: 状态={order.status}, 商品={', '.join(order.items)}, 金额=¥{order.total}"
    return f"未找到订单{order_id}"

@tool
def check_stock(product_name: str) -> str:
    """检查商品库存"""
    stock = {
        "iPhone 15": 120,
        "MacBook Pro": 45,
        "AirPods": 200,
    }
    quantity = stock.get(product_name, 0)
    return f"{product_name}: {'有货' if quantity > 0 else '缺货'} (库存{quantity}件)"

@tool
def get_customer_info(phone: str) -> str:
    """根据电话号码获取客户信息"""
    customers = {
        "13800138000": CustomerInfo(name="张三", member_level="金卡", phone="13800138000"),
        "13900139000": CustomerInfo(name="李四", member_level="银卡", phone="13900139000"),
    }
    customer = customers.get(phone)
    if customer:
        return f"客户: {customer.name}, 会员等级: {customer.member_level}"
    return "未找到客户信息"

@tool
def create_return_request(order_id: str, reason: str) -> str:
    """创建退货申请"""
    return f"已为订单{order_id}创建退货申请，原因：{reason}。工作人员将在24小时内联系您。"

# 3. 创建带Memory的Agent
tools = [query_order, check_stock, get_customer_info, create_return_request]

# 创建提示模板
prompt = ChatPromptTemplate.from_messages([
    ("system", """你是一个专业、友好的电商客服助手。

你的职责：
1. 礼貌地与客户沟通
2. 使用工具查询订单、库存等信息
3. 记住客户信息，提供个性化服务
4. 帮助客户解决问题

可用工具：
- query_order: 查询订单
- check_stock: 查询库存
- get_customer_info: 获取客户信息
- create_return_request: 创建退货申请

记住：始终保持专业和友好！"""),
    MessagesPlaceholder(variable_name="chat_history", optional=True),
    ("user", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# 创建Agent
agent = create_react_agent(llm=openai_model, tools=tools, prompt=prompt)

# 添加Memory
memory = ConversationBufferMemory(memory_key="chat_history", return_messages=True)

# 创建执行器
agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    memory=memory,
    verbose=False,
    handle_parsing_errors=True,
)

# 4. 模拟对话
print("\n" + "="*60)
print("客服对话演示")
print("="*60)

conversations = [
    "你好，我想查询一下订单ORD001的状态",
    "我还想知道iPhone 15还有货吗？",
    "我的电话是13800138000",
    "我想退货订单ORD001，商品有质量问题",
]

for i, user_msg in enumerate(conversations, 1):
    print(f"\n【第{i}轮对话】")
    print(f"👤 客户: {user_msg}")
    
    response = agent_executor.invoke({"input": user_msg})
    print(f"🤖 客服: {response['output']}")

print("\n" + "="*60)
print("💡 本案例展示的核心技能：")
print("="*60)
print("1. ✅ 使用多个专业工具（订单、库存、客户信息）")
print("2. ✅ Agent自主决策何时调用哪个工具")
print("3. ✅ Memory记住对话上下文（知道客户是谁）")
print("4. ✅ 结构化数据模型（CustomerInfo, OrderInfo）")
print("5. ✅ 专业的系统提示（定义客服角色）")
print("6. ✅ 可以轻松添加Streaming实现实时响应")
print("7. ✅ 可以添加Middleware实现日志、监控等功能")

print("\n🎓 这就是LangChain v1.0的强大之处！")
````

## 第十六章：综合实战案例 - 企业级知识问答系统

### 12.1 项目需求

构建一个完整的企业级知识问答系统，整合**所有核心组件**：

**功能需求：**
- 📚 支持多个知识库（产品、技术、政策）
- 💬 多轮对话，记住上下文
- 🔧 多种工具（搜索、计算、时间等）
- 📊 结构化输出查询结果
- 🎯 实时流式显示答案
- 📝 日志记录所有交互
- 🛡️ 敏感信息过滤

### 12.2 系统架构

```
用户输入
    ↓
[Middleware: 日志/过滤]
    ↓
[Agent: 智能决策]
    ↓
┌─────────────────────────────┐
│ Tools:                       │
│ - search_product (产品知识)  │
│ - search_tech (技术文档)     │
│ - calculator (计算器)        │
│ - get_time (时间)           │
└─────────────────────────────┘
    ↓
[Memory: 记住历史]
    ↓
[Streaming: 实时输出]
    ↓
[Structured Output: 格式化]
    ↓
用户得到答案
```

````python
import time
import json
from datetime import datetime
from typing import List, Optional
from pydantic import BaseModel, Field
from langchain.tools import tool
from langchain.agents import create_react_agent, AgentExecutor
from langchain.memory import ConversationBufferWindowMemory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings

print("="*80)
print(" "*20 + "🏢 企业级知识问答系统")
print("="*80)

# ========== 1. 定义结构化输出模型 ==========
class SearchResult(BaseModel):
    """搜索结果"""
    question: str = Field(description="用户问题")
    source: str = Field(description="信息来源")
    answer: str = Field(description="答案内容")
    confidence: float = Field(description="置信度0-1", ge=0, le=1)

# ========== 2. 准备多个知识库 ==========
print("\n📚 步骤1: 初始化知识库")
print("-" * 80)

# 产品知识库
product_docs = [
    "我们的旗舰产品SmartAI助手支持语音识别、自然语言处理和多轮对话。",
    "SmartAI助手月费为199元，年费1999元，企业版定制价格。",
    "SmartAI支持中文、英文、日语、韩语等15种语言。",
    "产品提供7天无理由退款，支持在线客服和电话技术支持。",
]

# 技术文档知识库
tech_docs = [
    "API接口使用RESTful规范，需要在请求头中携带API Key。",
    "系统采用微服务架构，支持水平扩展，日处理能力超过100万次请求。",
    "数据采用AES-256加密存储，通过ISO27001安全认证。",
    "SDK支持Python、Java、JavaScript、Go等主流编程语言。",
]

# 创建向量存储
embeddings = OpenAIEmbeddings()
product_vectorstore = FAISS.from_texts(product_docs, embeddings)
tech_vectorstore = FAISS.from_texts(tech_docs, embeddings)

print("✅ 产品知识库: 4条文档")
print("✅ 技术文档库: 4条文档")

# ========== 3. 定义工具集 ==========
print("\n🔧 步骤2: 定义工具集")
print("-" * 80)

@tool
def search_product_knowledge(query: str) -> str:
    """搜索产品知识库，了解产品功能、价格、支持等信息。"""
    docs = product_vectorstore.similarity_search(query, k=2)
    result = "\n".join(doc.page_content for doc in docs)
    return f"[产品知识] {result}"

@tool
def search_tech_docs(query: str) -> str:
    """搜索技术文档，了解API、架构、安全等技术信息。"""
    docs = tech_vectorstore.similarity_search(query, k=2)
    result = "\n".join(doc.page_content for doc in docs)
    return f"[技术文档] {result}"

@tool
def calculate(expression: str) -> str:
    """计算数学表达式。"""
    try:
        result = eval(expression)
        return f"计算结果: {expression} = {result}"
    except Exception as e:
        return f"计算错误: {str(e)}"

@tool
def get_current_time() -> str:
    """获取当前日期和时间。"""
    return datetime.now().strftime("%Y年%m月%d日 %H:%M:%S")

tools = [search_product_knowledge, search_tech_docs, calculate, get_current_time]
print(f"✅ 已加载 {len(tools)} 个工具")

# ========== 4. 创建日志中间件 ==========
print("\n📝 步骤3: 创建中间件")
print("-" * 80)

class LoggingMiddleware:
    """日志中间件"""
    def __init__(self):
        self.logs = []
    
    def log(self, event_type: str, data: dict):
        log_entry = {
            "timestamp": datetime.now().isoformat(),
            "type": event_type,
            "data": data
        }
        self.logs.append(log_entry)
        print(f"📝 [{event_type}] {datetime.now().strftime('%H:%M:%S')}")
    
    def get_logs(self):
        return self.logs

logger = LoggingMiddleware()
print("✅ 日志中间件已初始化")

# ========== 5. 创建Agent ==========
print("\n🤖 步骤4: 创建智能Agent")
print("-" * 80)

# 创建Memory
memory = ConversationBufferWindowMemory(
    memory_key="chat_history",
    return_messages=True,
    k=3  # 只保留最近3轮对话
)

# 创建提示模板
prompt = ChatPromptTemplate.from_messages([
    ("system", """你是一个专业的企业AI助手，名叫"小智"。

你的能力：
1. 搜索产品知识库 - 了解产品功能、价格、语言支持等
2. 搜索技术文档库 - 了解API、架构、安全等技术信息
3. 执行数学计算
4. 获取当前时间

工作准则：
- 友好专业，用"您"称呼用户
- 遇到不确定的问题，优先使用搜索工具
- 如果知识库中没有答案，诚实地告知
- 可以多次使用不同工具获取完整信息
- 记住对话历史，提供连贯的回答

请认真思考，为用户提供准确有价值的帮助！"""),
    MessagesPlaceholder(variable_name="chat_history", optional=True),
    ("user", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# 创建Agent
agent = create_react_agent(llm=openai_model, tools=tools, prompt=prompt)

agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    memory=memory,
    verbose=True,
    max_iterations=5,
    handle_parsing_errors=True,
)

print("✅ Agent创建成功")

# ========== 6. 包装执行函数（添加中间件） ==========
def execute_with_middleware(user_input: str) -> dict:
    """带中间件的执行"""
    # 记录输入
    logger.log("USER_INPUT", {"input": user_input})
    
    start_time = time.time()
    
    try:
        # 执行Agent
        result = agent_executor.invoke({"input": user_input})
        
        elapsed = time.time() - start_time
        
        # 记录输出
        logger.log("AGENT_OUTPUT", {
            "output": result["output"],
            "elapsed_time": f"{elapsed:.2f}s"
        })
        
        return result
    
    except Exception as e:
        logger.log("ERROR", {"error": str(e)})
        raise

# ========== 7. 运行测试对话 ==========
print("\n" + "="*80)
print(" "*25 + "💬 开始对话测试")
print("="*80)

test_conversations = [
    "你好，请介绍一下你自己",
    "你们的SmartAI助手支持哪些语言？",
    "价格是多少？如果买年费能便宜多少？",
    "你们的API是什么架构？",
    "现在几点了？",
]

for i, user_msg in enumerate(test_conversations, 1):
    print(f"\n{'='*80}")
    print(f"第 {i} 轮对话")
    print(f"{'='*80}")
    print(f"\n👤 用户: {user_msg}")
    print()
    
    result = execute_with_middleware(user_msg)
    
    print(f"\n🤖 小智: {result['output']}")
    
    time.sleep(1)  # 避免频繁调用

# ========== 8. 展示系统统计 ==========
print("\n" + "="*80)
print(" "*25 + "📊 系统统计报告")
print("="*80)

logs = logger.get_logs()
print(f"\n总交互次数: {len([l for l in logs if l['type'] == 'USER_INPUT'])}")
print(f"总日志条目: {len(logs)}")
print(f"\n最近5条日志:")
for log in logs[-5:]:
    print(f"  [{log['timestamp']}] {log['type']}")

# 内存统计
print(f"\n💾 Memory状态:")
print(f"  保留轮数: {memory.k}")
memory_vars = memory.load_memory_variables({})
if 'chat_history' in memory_vars:
    print(f"  当前历史: {len(memory_vars['chat_history'])} 条消息")

print("\n" + "="*80)
print("🎓 本案例整合的核心组件:")
print("="*80)
print("""
✅ Models          - GPT-4o作为核心大脑
✅ Messages        - 标准化的对话消息流
✅ Prompts         - 专业的系统提示模板
✅ Tools           - 4个专业工具（2个RAG + 2个功能）
✅ Agents          - ReAct Agent自主决策
✅ RAG             - 双知识库（产品+技术）
✅ Structured Out  - Pydantic定义SearchResult
✅ Memory          - Window Memory保留上下文
✅ Middleware      - 日志记录中间件
✅ Streaming       - 可选启用实时输出
""")

print("🚀 这就是一个完整的企业级AI系统！")
print("="*80)
````

### 8.4 错误处理与重试

````python
from tenacity import retry, stop_after_attempt, wait_exponential

# 使用重试装饰器
@retry(
    stop=stop_after_attempt(3),
    wait=wait_exponential(multiplier=1, min=2, max=10)
)
def call_model_with_retry(prompt: str):
    """带重试机制的模型调用。"""
    try:
        return openai_model.invoke(prompt)
    except Exception as e:
        print(f"调用失败：{e}，正在重试...")
        raise

# 测试
try:
    result = call_model_with_retry("你好")
    print("调用成功：", result.content)
except Exception as e:
    print(f"最终失败：{e}")
````

### 10.4 常见问题与解决方案

**Q1: Token消耗太大怎么办？**
