## 🧠***第二章 Model 模型***

- 模型是智能体的推理引擎。他们主导着客服的决策过程，决定调用哪些工具，如何解读结果，以及何时给出最终答案。

1. 与agent一起使用 - 模型可以在创建代理时动态指定。
2. 独立使用 -  模型可以直接（在agent循环之外）用于文本生成、分类或提取等任务，无需agent框架

### 初始化模型

在 LangChain 中开始使用独立模型最简单的方法是使用 init_chat_model 从你选择的聊天模型提供商中初始化一个模型（示例如下）：


````python
import os
from langchain.chat_models import init_chat_model

os.environ["DEEPSEEK_API_KEY"] = "sk-b4c344ce9f444cd59b23a6dea522a902"

llm_deepseek = init_chat_model(
    # 可以设置一下参数等
    model="deepseek-chat", # deepseek-reasoner
    api_key=os.environ["DEEPSEEK_API_KEY"],
    base_url="https://api.deepseek.com", 
    temperature=0.7, # 控制模型输出的随机性。数字越高，回答越有创意;较低的分数使得它们更具确定性
    max_tokens=1000, # 最大token数,在响应中，有效控制输出的长度。
    timeout=60, # 超时，在取消请求前等待模型响应的最长时间（以秒为单位）。
    max_retries=3 # 如果请求因网络超时或速率限制等问题失败，系统最多尝试重发请求的次数。
)

response = llm_deepseek.invoke("你好，介绍一下你自己！")

print(response.content)
````

## ♥️***第三章 创建第一个agent***

- 智能体将语言模型与工具结合起来，创建能够推理任务、决定使用哪些工具并迭代寻找解决方案的系统。
- 一个 LLM Agent 会循环运行工具以实现目标。智能体运行直到满足停止条件——即模型输出最终输出或达到迭代极限。

![agent1.png](./image/agent1.png)

````python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """ 获取一个给定城市的天气 """
    return f"今天{city}是晴天！"

agent = create_agent(
    model=llm_deepseek, # 模型
    tools=[get_weather], # 工具
    system_prompt="你是一个天气助手", # 系统提示
)

# Run the agent
agent.invoke(
    {"messages": [{"role": "user", "content": "今天上海的天气如何？"}]}
)

````

## 🪧***第四章 Message 消息***

`消息是 LangChain 模型的基本上下文单位。它们代表模型的输入和输出，携带表示与大型语言模型交互对话状态所需的内容和元数据。`

消息是包含以下内容的对象：
- role 角色 - 标识消息类型(例如，system、user)
- content 内容 - 表示消息的实际内容（如文本、图片、音频、文档等）
- metadata 元数据 - 可选字段，如响应信息、消息 ID 和令牌使用情况

### **基本用法**

`使用消息的最简单方式是创建消息对象，并在调用时传递给模型。`


````python
from langchain.messages import HumanMessage, AIMessage, SystemMessage

system_msg = SystemMessage("你是一个感情受伤的小男孩。")
human_msg = HumanMessage("你好，你怎么样？")

# Use with chat models
messages = [system_msg, human_msg]
response = llm_deepseek.invoke(messages)  # Returns AIMessage

print(response.content)
````

#### AI Message  AI 信息

`AIMessage 表示模型调用的输出。它们可以包括多模态数据、工具调用以及你之后可以访问的提供者特定元数据。`

`AIMessage 对象在调用模型时返回，包含响应中所有相关的元数据。`

````python
response = llm_deepseek.invoke("Explain AI")
print(type(response))  # <class 'langchain.messages.AIMessage'>
````

`提供者对消息类型的权重和语境不同，这意味着有时手动创建一个新的 AIMessage 对象并将其插入消息历史中，就像它来自模型一样。`

````python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# Create an AI message manually (e.g., for conversation history)
ai_msg = AIMessage("I'd be happy to help you with that question!")

# Add to conversation history
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,  # Insert as if it came from the model
    HumanMessage("Great! What's 2+2?")
]

response = llm_deepseek.invoke(messages)

print(response)
````

Tool calls  工具调用

`当模型调用工具时，这些调用会被包含在 AIMessage 中：`

````python
from langchain.chat_models import init_chat_model


def get_weather(location: str) -> str:
    """Get the weather at a location."""
    ...

model_with_tools = llm_deepseek.bind_tools([get_weather])
response = model_with_tools.invoke("What's the weather in Paris?")

for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
    print(f"ID: {tool_call['id']}")
````

#### Tool Message  工具信息

`对于支持工具调用的模型，AI 消息可以包含工具调用。工具消息用于将单个工具执行的结果反馈给模型。`

工具可以直接生成 ToolMessage 对象。

````python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72°F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,  # Model's tool call
    tool_message,  # Tool execution result
]
response = llm_deepseek.invoke(messages)  # Model processes the result

print(response.content)
````

## 🛠️***第五章 Tools 工具***

`许多人工智能应用通过自然语言与用户互动。然而，有些用例需要模型通过结构化输入直接与外部系统（如 API、数据库或文件系统）接口。`

`工具是agent调用以执行动作的组件。它们通过明确定义的输入输出与世界互动，扩展了模型能力。`

`工具封装了一个可调用函数及其输入模式。这些参数可以传递给兼容的聊天模型 ，使模型决定是否调用某个工具以及使用何种参数。在这些场景中，工具调用使模型能够生成符合指定输入模式的请求。`


### 工具的定义

`制作工具最简单的方式是用 @tool 装饰器。默认情况下，函数的文档字符串成为工具的描述，帮助模型了解何时使用：`

````python
from langchain.tools import tool

@tool
def search_database(query: str, limit: int = 10) -> str:
    """在客户数据库中搜索与查询匹配的记录。

    Args:
        query: 要查找的搜索词
        limit: 要返回的最大结果数量
    """
    return f"Found {limit} results for '{query}'"
````

类型提示是必需的，因为它们定义了工具的输入模式。文档字符串应具备信息性且简洁，帮助模型理解工具的目的。

##### 自定义工具属性

自定义工具名称与描述


````python
@tool("calculator", description="Performs arithmetic calculations. Use this for any math problems.")
def calc(expression: str) -> str:
    """Evaluate mathematical expressions."""
    return str(eval(expression))
````

## ⚙️***Middleware 中间件***

`在每一步控制并定制agent执行`

中间件提供了一种更严格控制agent内部发生事情的方法。中间件适用于以下用途：

- 通过日志记录、分析和调试跟踪代理行为。
- 转换提示、 工具选择和输出格式。
- 增加了重试 、 后备和提前终止逻辑。
- 应用速率限制 、保护栏和个人身份识别（PII）检测 。


通过传递给 create_agent 添加中间件：

````python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware, HumanInTheLoopMiddleware

agent = create_agent(
    model=llm_deepseek,
    tools=[...],
    middleware=[
        SummarizationMiddleware(...),
        HumanInTheLoopMiddleware(...)
    ],
)
````

### 内置中间件

`LangChain 为常见场景提供了预构建的中间件。`

| 中间件 | 描述|
| ----- | -----|
|Summarization  综述 | 接近代币限制时自动总结对话记录。|
|Human-in-the-loop  人机交互	| 暂停执行以供人工批准工具调用。|
|Model call limit  模型呼叫限制	| 限制模型调用次数，以防止过高成本。|
|Tool call limit  工具调用限制	| 通过限制呼叫次数来控制工具执行。|
|Model fallback  模型的备选	| 当主模式失败时，会自动回退到其他模式。|
|PII detection  PII 检测	| 检测并处理个人身份信息（PII）。|
|To-do list  待办事项列表	| 为客服人员配备任务规划和跟踪能力。|
|LLM tool selector  LLM工具选择器	| 在调用主模型之前，先用 LLM 选择相关工具。|
|Tool retry  工具重试	| 用指数回撤自动重试失败的工具调用。|
|Model retry  模型重试	| 自动用指数退回方式重试失败的模型调用。|
|LLM tool emulator  LLM工具仿真器	| 用 LLM 模拟工具执行以进行测试。|
|Context editing  上下文编辑	| 通过修剪或清理工具使用来管理对话上下文。|
|Shell tool  shell工具	| 向代理开放一个持久的shell会话以执行命令。|
|File search  文件搜索	| 在文件系统文件上提供 Glob 和 Grep 搜索工具。|
